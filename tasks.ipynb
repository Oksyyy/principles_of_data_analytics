## Task 1: Source the Data Set

### Import sklearn.dataset package

The sklearn.datasets package embeds several toy datasets. These datasets don't require downloading external files and are often used for practicing machine learning techniques. 

```python
# Import prebuilt load_iris()function, which enables loading Iris dataset

from sklearn.datasets import load_iris
```

### Load dataset

load_iris() is a prebuilt function within scikit-learn library, that loads a famous Iris dataset. This dataset contains data about 3 species of Iris flower with 4 flower measurments. 

The load_iris() function returns a Bunch object containing: 
- data: a table (array) with 150 records and 4 columns representing sepals and petals measurments of Iris flowers
- target: numeric values representing 3 classes of Iris species (0,1,2)
- target_names: names of each species class (setosa, versicolor, virginica)
- feature_names: names of measurements (sepal length (cm), sepal width (cm), petal length (cm), petal width (cm))
- DESCR: text description of the Iris dataset

```python
# Load data
iris = load_iris()
```

```python
# Populate data

print(iris)
```

```python
# Return data keys of the data set (diferent parts of the data set of what's loaded into memory)

iris.keys()
```

```python
# See dataset description that provides characteristics and statistics summary

print(iris.DESCR)
```

## Task 2: Explore the Data Structure

.shape returns dimentions of the data set, such as the number of rows and columns, in a tuple format. 

```python
# Return shape of the data set
# Iris data set consists of 150 rows and 4 columns

iris['data'].shape
```

For Iris data set .shape can also be used to return a single variable

```python
# Return dimensions of the target variable 
# Iris dataset contains 150 target values

iris['target'].shape
```

Use NumPy Array Slicing [start:end] to return the first & the last 5 rows of the data set

Resource: https://www.w3schools.com/python/numpy/numpy_array_slicing.asp

```python
# Return the first 5 rows of the data set
# If start is not passed, it's considered 0

iris.data[:5]
```

```python
# Return the last 5 rows of the data set
# If end is not passed, it's considered until the last element in an array

iris.data[-5:]
```

```python
# Return feature names
# Feature names correspond to varios measurments of Iris flowers

print(iris.feature_names)
```

```python
# Return target clases 
# The data set contains 3 distinct numeric target classes corresponding to 3 Iris species

print(iris.target)
```
