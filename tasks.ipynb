## Task 6: Analyze Relationship

To better demonstrate the relationship between sepal length and and petal length in the Iris dataset, we will add a regression line to the scatter plot created in Task 5. 

We will use [numpy.polyfit()](https://numpy.org/doc/stable/reference/generated/numpy.polyfit.html) function, which helps to find a line, that best fits our data points. This method requires three arguments: x, y and the degree of polynomial. For a straight line we will use a degree = 1. 

Under the hood, numpy.polyfit() uses the ordinary least squares method, which minimizes the total squared distance between the actual data points and the values predicted by the line.

A helpful walkthrough of how linear regression works can be found in an article by [Data36](https://data36.com/linear-regression-in-python-numpy-polyfit/).

```python
# Define x, y variables
x = sepal_length
y = petal_length
```

```python
# Use polyfit to fit a line to the dataset
# "m" is the slop of the line (how steep the line is)
# "c" is the intercept (where the line crosses y axes)
# 1 is the degree (1 = straight line)
m, c = np.polyfit(x, y, 1)

# Show m & c
m, c
```

```python
# Create scatterplot
fig, ax = plt.subplots()

# Plot the data 
ax.scatter(x, y, marker =  'v',  c = target_class)

# Set axes labels
ax.set_xlabel('Sepal Length (cm)')
ax.set_ylabel('Petal Length (cm)')

# Fit the regression line
ax.plot(x, m * x + c, color = 'red', label = 'Regression line')

# Add legend
ax.legend()

# Add title 
ax.set_title('Sepal Length and Petal Length Relationship by Iris Species')
```

## Task 8: Compute Correlations

We will compute correlation between Iris features using **Pearson correlation coefficient**, which is: 
>A measure of the linear relationship between two features. It’s the ratio of the covariance of x and y to the product of their standard deviations. It’s often denoted with the letter r and called Pearson’s r.

(Source: [RealPython](https://realpython.com/numpy-scipy-pandas-correlation-python/#pearson-correlation-coefficient) tutorial)

The Pearson correlation coefficient ranges between -1 and 1 and can be interpreted as follows: 

| Coefficient |                          Relationship|
|:-----------:|:------------------------------------:|
| 1           | perfect positive linear relationship |
| 0           | no linear relationship               |
| -1          | perfect negative linear relationship |

There are various ways to compute Pearson correlation coefficient in Python as, discussed in the linked tutorial. 

For this task, We will use [numpy.corrcoef](https://numpy.org/doc/2.2/reference/generated/numpy.corrcoef.html) function.

```python
# Generate numpy array 
# The first ":" selects all rows
# The second ":4" selects the first 4 columns 
# We explude the class column, as it is categorical and not suitable for correlation calculation
x = iris.data[:,:4]
```

```python
# Calculate correlation matrix using np.corrcoef() function
# Set rowvar=False to treat columns as variable (features) and rows as observations
correlation = np.corrcoef(x, rowvar=False)

# Show correlation
correlation
```

### Visualize Rlationship Between Iris Features - Create a Heatmap
We will use [Annotated Heatmap](https://matplotlib.org/stable/gallery/images_contours_and_fields/image_annotated_heatmap.html#sphx-glr-gallery-images-contours-and-fields-image-annotated-heatmap-py) from Matplotlib module and [Colormaps](https://matplotlib.org/stable/users/explain/colors/colormaps.html) for color palet selection.

```python
# Plot figure and axes
fig, ax = plt.subplots()

# Set x & y axes labels
x = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
y = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

# Generate a heatmap using matplotlib imshow() function
# Use cmap='PRGn' to set plot colors for better interpretation of results
ax.imshow(correlation, cmap='PRGn')

# Set custom ticks and label them with the respective feature names
# Use range() to generate a sequence of positions for ticks (and label them as features)
# The use of len() gives the total number of features
# The argument ha="right" aligns the right edge of label text with the tick mark on x axis 
# Rotation by 45 degrees improves readability of labels
ax.set_xticks(range(len(x)), labels=x, rotation=45, ha="right")
ax.set_yticks(range(len(y)), labels=y)

# Loop over data dimensions and create text annotations
# The iteration structure is used from one of the plotting examples shown in the documentation linked above
# F-string and ":.2f" are used to format and round values to 2 decimal places for clarity
for i in range(len(y)):
    for j in range(len(x)):
        text = ax.text(j, i, f"{correlation[i, j]:.2f}",
                       ha="center", va="center", color="w")
# Set title
ax.set_title("Correlation between Iris measurements ")
```

The above plot tells us that:
- There's an almost perfect positive lenear relationship between Petal lenth and Petal width (0.96).
- There's almost no linear relationship between Sepal length and Sepal width (-0.12).
- There's a good relationship between Sepal length and Petal width (0.82).
- Each feature is always perfectly correlated with itself, i.e. Sepal length and Sepal length(1).

## Task 9: Fit a Simple Linear Regression

In this task, we will focus on evaluating how good the fit of the best fit line (created in Task 6) actually is. This can be done using the **Coefficient of determination**, or **$R^2$**.

>The coefficient of determination is the proportion of the variance in the response variable that can be explained by the predictor variable. It can range from 0 to 1. 
>A value of 0 indicates that the response variable cannot be explained by the predictor variable at all. A value of 1 indicates that the response variable can be perfectly explained without error by the predictor variable.

(Source: [Statology.org](https://www.statology.org/linear-regression/))

### Scipy

For calculating $R^2$ in Python we will use [scipy](https://docs.scipy.org/doc/scipy/tutorial/index.html#user-guide) module. Scipy is built on top of numpy and provides a collection on mathematical algorithms and functions. 

Within Scipy, we will use the [linregress()](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.linregress.html) function to compute coefficient of determination.

```python
# Define x, y variables (same features as analyzed in task 5)
x = sepal_length
y = petal_length
```

```python
# Use scipy to calculate the best fit parameters for fitting a line
fit = sp.stats.linregress(x,y)

# Show
fit
```

```python
# Slope
print(f"Slope:        {fit.slope}")

# Intercept
print(f"Intercept:    {fit.intercept}")

# R value
print(f"R value:      {fit.rvalue}")

# R squared
print(f"R squared:    {fit.rvalue**2}")
```

```python
# Fit a line using numpy polyfit (used earlier in Task 6)
# "m" is the slope of the line (how steep the line is)
# "c" is the intercept (where the line crosses y axes)
# 1 is the degree (1 = straight line)
m, c = np.polyfit(x, y, 1)

# Show slope and intercept
m, c
```

We can see that the slope and intercept values created using linregress() and polyfit() are very close, with only minor differences due to rounding and internal calculation methods.

Both methods apply least squares fitting.

```python
# Create scatterplot (re-creating the plot from Task 6)
fig, ax = plt.subplots()

# Plot the data 
ax.scatter(x, y, marker =  'v',  c = target_class)

# Set axes labels
ax.set_xlabel('Sepal Length (cm)')
ax.set_ylabel('Petal Length (cm)')

# Fit the regression line
# Use the slope and intercept from sp.stats.linregress
# Add R^2 value to the plot
# Use f-string to format label and use ":.2f" to round R^2 to 2 decimal places for clarity
ax.plot(x, fit.slope * x + fit.intercept, color = 'red', label = f"$R^2$ = {fit.rvalue**2:.2f}")

# Add legend
ax.legend()

# Add title 
ax.set_title('Sepal Length and Petal Length Relationship by Iris Species')
```

The **$R^2$** value of **0.76** (rounded) indicates that approximately **76%** of the variation in Petal Length can be explained by the variation in Sepal length.
