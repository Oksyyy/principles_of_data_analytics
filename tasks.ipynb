## Task 1: Source the Data Set

### Scikit-Learn

This project is using Scikit-Learn library to load and analyse a dataset. As part of the sklearn.datasets package, the library includes a number of toy datasets, which don't require downloading external files. These are often used for practicing machine learning techniques: https://scikit-learn.org/stable/datasets/toy_dataset.html

The load_iris() function is a prebuilt method within Scikit-learn that loads the well-known Iris dataset. It contains measurements for 3 Iris flower species. 

The load_iris() function returns a Bunch object containing: 
- data: a table (array) with 150 records and 4 columns representing sepals and petals measurements of Iris flowers
- target: numeric values representing 3 classes of Iris species (0,1,2)
- target_names: names of each species class (setosa, versicolor, virginica)
- feature_names: names of measurements (sepal length (cm), sepal width (cm), petal length (cm), petal width (cm))
- DESCR: text description of the Iris dataset

```python
# Load data
iris = load_iris()
```

```python
# Populate data

print(iris)
```

```python
# Return data keys of the data set (different parts of the data set of what's loaded into memory)

iris.keys()
```

```python
# See dataset description that provides characteristics and statistics summary

print(iris.DESCR)
```

## Task 5: Investigate Relationships

In this task, we will investigate the relationship between sepal length and petal length using a scatter plot. Scatter plots help to visualise how two variables are related. We will use: 
- Numpy package for accessing and generating data arrays,
- Matplotlib library for plotting

```python
# Keys
iris.keys()
```

```python
# Show feature_names
iris.feature_names
```

```python
# Sepal length - will be treated as x axes
sepal_length = iris.data[:,0]

# Show first 10 
sepal_length[:10]
```

```python
# Petal length - will be treated as y axes
petal_length = iris.data[:,2]

# Show first 10 
petal_length[:10]
```

```python
# Show unique Iris species classes (target variable)
# 0 = Setosa, 1 = Versicolor, 2 = Virginica
target_class = iris.target

# Show first 10 values of species classes
target_class[:10]
```

```python
# Create a figure and axes using subplots, which provide more controle over plots elements
# Referrred to Matplotlib ducumentation: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html
fig, ax = plt.subplots()

# Scatter plot of Sepal Length vs Petal Length  
# "marker" v sets values figures to triangles
# "c" argument assigns colours by class 
ax.scatter(sepal_length, petal_length, marker =  'v',  c = target_class)

# Set axes labels
ax.set_xlabel('Sepal Length (cm)')
ax.set_ylabel('Petal Length (cm)')
```

## Task 7: Analyze Class Distributions

```python
# List unique class labes in the dataset using np.unique()
# Class labels correspond to different Iris species: 0 = Setosa, 1 = Versicolor, 2 = Virginica
x = np.unique(iris['target'])

# Display unique labels
x
```

```python
# Extract petal length for class 1 (Setosa) 
# From previous analysis, petal_length is the third columnn in the dataset (index = 2)
# Select rows where the target value is 0 (Setosa) and extract petal length columnn 
class_1 = iris.data[iris.target == 0, 2]
```

```python
# Extract petal length for class 2 (Versicolor) 
# Select rows where the target value is 1 (Versicolor) and extract petal length columnn 
class_2 = iris.data[iris.target == 1, 2]
```

```python
# Extract petal length for class 3 (Virginica) 
# Select rows where the target value is 2 (Virginica) and extract petal length columnn 
class_3 = iris.data[iris.target == 2, 2]
```

### Organize data

```python
# Create a numpy array to store petal length values for each class
data = np.array([class_1, class_2, class_3])

# Display petal length dataset
print(data)
```

### Visualaze Petal Length Using Boxplots

```python
# Create figure and axis for the plot
fig, ax = plt.subplots()

# Create boxplots
# Transpose the data using data.T allowing for each class to create a separate boxplot
ax.boxplot(data.T)

# Add title, axes labels
ax.set_title('Petal Length By Species Class Boxplots', fontsize=16)
ax.set_xlabel('Species', fontsize=12)
ax.set_ylabel('Petal Length cm', fontsize=12)

# Add names of x axis ticks
ax.set_xticks([1,2,3], ["Setosa", "Versicolor", "Virginica"], fontsize=10)

# Add a grid
ax.grid(axis="y", linestyle="--", alpha=0.5)
```
